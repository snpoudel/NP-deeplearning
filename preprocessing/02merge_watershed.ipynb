```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
import numpy as np
```

```python
# read list of basins
basins_list = pd.read_csv('rawdata/web_based/catchment_areas.csv')
# reac each geojson file, to each file add attribute called gauge_id which is the id number of the basin, then merge all geojson files into one geodataframe
gdf_list = []
for id in basins_list['basin_id']:
    gdf = gpd.read_file(f'rawdata/web_based/{id}.geojson')
    #only_id =  # extract only the number from the id string
    only_id = float(id.split('_')[1])
    gdf['gauge_id'] = only_id
    gdf_list.append(gdf)
gdf_all = pd.concat(gdf_list, ignore_index=True)
# gdf_all.to_file('rawdata/web_based/catchment_areas.geojson', driver='GeoJSON')

# save merged file as a shapefile
gdf_all.to_file('rawdata/web_based/merged_shapefile.shp', driver='ESRI Shapefile')
# plot the merged shapefile
gdf_all.plot(column='gauge_id', cmap='viridis', legend=True)
```
